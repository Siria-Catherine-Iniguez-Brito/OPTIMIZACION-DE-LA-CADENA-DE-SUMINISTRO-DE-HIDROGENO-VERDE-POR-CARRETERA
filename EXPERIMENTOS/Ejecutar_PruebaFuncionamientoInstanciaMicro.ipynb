# Ejecutar prueba_funcionamiento_instancia_micro.py en Colab

Este notebook monta tu Google Drive, localiza el script
`prueba_funcionamiento_instancia_micro.py` dentro de tu carpeta
`EXPERIMENTOS/`, ajusta sus rutas internas (que en el script original son
absolutas de tipo `/TFM/...`, pensadas para ejecucion LOCAL) para que
apunten a la ubicacion real dentro de tu Drive, y lo ejecuta sin necesidad
de instalar nada.

No hace falta copiar el codigo del script aqui: el notebook lo importa
directamente desde donde lo tengas guardado en Drive.


```python
# =====================================================================
# CELDA 1 - Montar Drive y localizar la carpeta EXPERIMENTOS/
# =====================================================================
import os, sys

DRIVE_OK = False
try:
    from google.colab import drive
    drive.mount("/content/drive", force_remount=False)
    DRIVE_OK = True
except Exception as e:
    print("[INFO] No se pudo montar Drive (fuera de Colab?):", e)

# Buscamos la carpeta EXPERIMENTOS/ (la que contiene el script) recorriendo
# todo el Drive montado.
NOMBRE_SCRIPT = "prueba_funcionamiento_instancia_micro.py"
raiz_busqueda = "/content/drive" if os.path.isdir("/content/drive") else "/content"

candidatos_script = []
for raiz, _dirs, ficheros in os.walk(raiz_busqueda):
    if NOMBRE_SCRIPT in ficheros:
        candidatos_script.append(os.path.join(raiz, NOMBRE_SCRIPT))
candidatos_script.sort()

if not candidatos_script:
    raise FileNotFoundError(
        f"No se encontro '{NOMBRE_SCRIPT}' dentro de {raiz_busqueda}. "
        "Comprueba que esta subido a tu Drive dentro de la carpeta EXPERIMENTOS."
    )

RUTA_SCRIPT = candidatos_script[0]
DIR_EXPERIMENTOS = os.path.dirname(RUTA_SCRIPT)
print("Script encontrado en:", RUTA_SCRIPT)
if len(candidatos_script) > 1:
    print("  [AVISO] Hay varias copias del script; se usa la primera:", candidatos_script)

# Estructura esperada: .../TFM/{EXPERIMENTOS, CODIGO_FUENTE, DATOS/tablas}
DIR_TFM = os.path.dirname(DIR_EXPERIMENTOS)
print("Carpeta EXPERIMENTOS:", DIR_EXPERIMENTOS)
print("Carpeta TFM detectada:", DIR_TFM)

RUTA_REPO_REAL = os.path.join(DIR_TFM, "CODIGO_FUENTE")
RUTA_INSTANCIA_REAL = os.path.join(DIR_TFM, "DATOS", "Tablas", "instancia_micro.json")

for etiqueta, ruta in [("CODIGO_FUENTE", RUTA_REPO_REAL),
                       ("instancia_micro.json", RUTA_INSTANCIA_REAL)]:
    existe = os.path.isdir(ruta) if etiqueta == "CODIGO_FUENTE" else os.path.isfile(ruta)
    print(f"  {etiqueta:<22} {'OK' if existe else '[AVISO] no encontrado'} -> {ruta}")
```

```python
# =====================================================================
# CELDA 2 - Importar el script como modulo y AJUSTAR sus rutas para Colab
# =====================================================================
import importlib.util

spec = importlib.util.spec_from_file_location("prueba_micro", RUTA_SCRIPT)
prueba_micro = importlib.util.module_from_spec(spec)
spec.loader.exec_module(prueba_micro)

# El script trae rutas absolutas de tipo "/TFM/..." pensadas para ejecucion
# LOCAL fuera de Colab. Las sobreescribimos aqui para que apunten a la
# ubicacion real dentro de tu Google Drive, sin tocar el fichero original.
prueba_micro.CONFIG["RUTA_REPO"] = RUTA_REPO_REAL
prueba_micro.CONFIG["RUTA_INSTANCIA"] = RUTA_INSTANCIA_REAL
prueba_micro.CONFIG["DIR_EXPERIMENTO"] = os.path.join(
    DIR_EXPERIMENTOS, "PruebaFuncionamientoInstanciaMicro"
)

print("CONFIG ajustado para Colab:")
for k, v in prueba_micro.CONFIG.items():
    if k != "GA":
        print(f"  {k:<16} = {v}")
print("  GA =", prueba_micro.CONFIG["GA"])
```

```python
# =====================================================================
# CELDA 3 - Ejecutar el pipeline completo (main)
# =====================================================================
prueba_micro.main()
```
